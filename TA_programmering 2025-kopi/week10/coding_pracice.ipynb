### Code 10.2: Weather Station Revisited

```python
class WeatherStation:
    def __init__(self, station_name, is_operational):
        self.station_name = station_name
        self.is_operational = is_operational
        self.wind_readings = []
        self.temperature_readings = []
    
    def take_daily_wind_readings(self, wind_numbers):
        self.wind_numbers = wind_numbers

        if self.is_operational:
            for i in self.wind_numbers:
                self.wind_readings.append(i)
        else: 
            print('The station is not operational')
    
    def take_daily_temperature_readings(self, temp_numbers):
        self.temp_numbers = temp_numbers

        if self.is_operational:
            for i in self.temp_numbers:
                self.temperature_readings.append(i)
        else: 
            print('The station is not operational')
    
    def get_highest_readings(self):
        if len(self.temperature_readings) != 0 and len(self.wind_readings) != 0:
            return max(self.wind_readings), max(self.temperature_readings)
        elif len(self.temperature_readings) == 0 and len(self.wind_readings) == 0:
            return None, None
        elif len(self.temperature_readings) == 0 and len(self.wind_readings) != 0:
            return max(self.wind_readings), None
        else:
            return  None, max(self.temperature_readings)


    

cph_station = WeatherStation("Copenhagen", True)
print(cph_station.station_name)
#Copenhagen
print(cph_station.is_operational)
#True
#print(cph_station.wind_readings)
#[]
#print(cph_station.temperature_readings)
#[]

cph_station.take_daily_wind_readings([3.5, 6.2, 7.1, 12.1])
cph_station.take_daily_temperature_readings([18.1, 18.3, 19.0, 18.5])
print(cph_station.wind_readings)
#[3.5, 6.2, 7.1, 12.1]
print(cph_station.temperature_readings)
#[18.1, 18.3, 19.0, 18.5]

max_wind, max_temp = cph_station.get_highest_readings()
#print(max_wind)
#12.1
#print(max_temp)
#19.0
```

### Code 10.3: Health Analysis Revisited

```python
class PersonHealth:
    def __init__(self, age, height, weight, sex):
        self.age = age
        self.height = height
        self.weight = weight
        self.sex = sex

        if self.sex == 'm':
            self.sex = 1
        else:
            self.sex = 0
    
    def body_mass_index(self):
        BMI = self.weight / (self.height)**2
        return round(BMI, 2)
    
    def body_fat_percentage(self):
        BMI = self.body_mass_index()
        PBF = 1.2 * BMI + 0.23 * self.age - 10.8 * self.sex - 5.4
        return round(PBF, 2)

    def get_health_description(self):
        BMI = self.body_mass_index()
        PBF = self.body_fat_percentage()
        return f'The person is {self.age} years old, with at BMI {BMI} and PBF {PBF}' 



person1 = PersonHealth(20, 1.70, 69, 'f')
print(person1.get_health_description())
#'Age: 20, BMI: 23.88, BFP: 27.85'
person2 = PersonHealth(61, 1.92, 91, 'm')
print(person2.get_health_description())
#'Age: 61, BMI: 24.69, BFP: 27.45'
```
