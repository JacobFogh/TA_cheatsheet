### Problem 10.4: Simple Statistics

```python
import math
class SimpleStatistics:

    def __init__(self, samples):
        self.samples = samples

    def get_mean(self):
        return sum(self.samples)/len(self.samples)
    
    def get_standard_deviation(self):
        mean = self.get_mean()
        std = math.sqrt(sum([(i - mean)**2 for i in self.samples]) / (len(self.samples) - 1))
        return std
    
    def one_sample_ttest(self, mu_zero):
        mean = self.get_mean()
        std = self.get_standard_deviation()
        N = len(self.samples)
        t = (mean - mu_zero) / (std / math.sqrt(N))

        if abs(t) >= 1.96:
            return True
        else:
            return False 

    

samples = [0.1763, -1.2523, 0.9099, -0.0679, 0.1999, 0.1175, 0.143, 0.1118, 0.9854, -1.1836, -0.5868, 0.4839, 1.2069, 0.142, 0.2523]
stat = SimpleStatistics(samples)

print(stat.get_mean())
#0.1092200000000000

print(stat.get_standard_deviation())
#0.6984926241557602

print(stat.one_sample_ttest(0))
#False

print(stat.one_sample_ttest(0.1))
#False

print(stat.one_sample_ttest(-0.1))
#False

print(stat.one_sample_ttest(-0.25))
#True
```

### Problem 10.5: Study Planner

```python
class StudyPlanner:

    def __init__(self, blocks, points):
        self.blocks = blocks
        self.points = points

        self.registered_courses = {x : [] for x in blocks}
        self.points_left = {x : y for x, y in zip(blocks, points)}

    def add_course(self, course_name, course_block, ECTS):
        if self.points_left[course_block] >= ECTS:
            self.points_left[course_block] -= ECTS
            self.registered_courses[course_block].append(course_name)
            return 1
        else:
            return -1
    
    def print_courses(self):

        # for i in self.registered_courses:
        #     print(f"The course block {i.upper()} has {self.registered_courses[i]} courses with {self.points_left[i]} points left")

            print("---------------")
            for block in self.registered_courses:
                print(f"BLOCK {block.upper()}")
                print(f"Points left: {self.points_left[block]}")
                print("Courses:")
                for course in self.registered_courses[block]:
                    print(f"  {course}")

            print("---------------")
            
blocks = ['foundations', 'directions', 'projects', 'electives']
points = [55, 55, 25, 45]
planner = StudyPlanner(blocks, points)
#planner.print_courses()
planner.add_course('Programming', 'foundations', 5)
#planner.print_courses()
planner.add_course('Mathematics 1a', 'foundations', 10)
planner.add_course('Mathematics 1b', 'foundations', 10)
planner.add_course('Signals and systems', 'directions', 5)
planner.add_course('Optics and photonics', 'directions', 5)
planner.add_course('BSc project', 'projects', 15)
planner.add_course('Prototyping', 'projects', 10)
planner.add_course('Acoustics and noise', 'electives', 5)
planner.print_courses()
```

### Problem 10.6: Bank Account 

```python
class BankAccount:

    def __init__(self, balance):
        self.balance = balance

    def deposit(self, deposit_amount):
        self.balance += deposit_amount
    
    def withdraw(self, withdaw_amount):
        self.balance -= withdaw_amount

        if self.balance < 0:
            self.balance += withdaw_amount
            return 0
        else:
            return withdaw_amount
    
    def get_balance(self):
        return self.balance



my_account = BankAccount(1000)
print(my_account.get_balance())
#1000
my_account.deposit(500)

print(my_account.get_balance())
#1500
print(my_account.withdraw(200))
#200
print(my_account.get_balance())
#1300
print(my_account.withdraw(2000))
#0
print(my_account.get_balance())
#1300
```

### Problem 10.7: Event Manager

```python
class EventManager:

    def __init__(self):
        self.name_register = []
    
    def register(self, name):
        if name not in self.name_register:
            self.name_register.append(name)
            return True
        else:
            return False
    
    def deregister(self, name):
        if name in self.name_register:
            self.name_register.remove(name)
            return True
        else:
            return False
    
    def get_num_registrations(self):
        return len(self.name_register)

my_event = EventManager()
print(my_event.get_num_registrations())
#0
print(my_event.deregister('Mike'))
#False
print(my_event.register('Mike'))
#True
print(my_event.register('Mike'))
#False
print(my_event.register('John'))
#True
print(my_event.deregister('Mike'))
#True
print(my_event.get_num_registrations())
#1
```
