```python
class BankAccount:

    def __init__(self, balance):
        self.balance = balance

    def deposit(self, deposit_amount):
        self.balance += deposit_amount
    
    def withdraw(self, withdaw_amount):
        self.balance -= withdaw_amount

        if self.balance < 0:
            self.balance += withdaw_amount
            return 0
        else:
            return withdaw_amount
    
    def get_balance(self):
        return self.balance

```

```python
class OverdraftAccount(BankAccount):
    def __init__(self, balance, overdraft_limit):
        super().__init__(balance)
        self.overdraft_limit = overdraft_limit
    
    def withdraw(self, withdraw_amount):
        if self.balance - withdraw_amount < -self.overdraft_limit:
            return 0
        else:
            self.balance -= withdraw_amount
            return withdraw_amount
    
my_account = OverdraftAccount(0, 500)

my_account.deposit(1000)

print(my_account.get_balance())
#1000
print(my_account.withdraw(1300))
#1300
print(my_account.get_balance())
#-300
print(my_account.withdraw(500))
#0
print(my_account.get_balance())
#-300   
```
