### Code 11.2: Score Tracker

```python
class ScoreTracker:
    def __init__(self):
        self.score_1 = 0
        self.score_2 = 0
        self.name_1 = ''
        self.name_2 = ''
    
    def include(self,new_score, new_name):
        self.new_name = new_name
        self.new_score = new_score

        if self.new_score >= self.score_1:
            self.name_2 = self.name_1
            self.name_1 = self.new_name

            self.score_2 = self.score_1
            self.score_1 = self.new_score

        elif self.new_score >= self.score_2:
            self.name_2 = self.new_name
            self.score_2 = self.new_score

    
    def __str__(self):
        return f'High scores \nwinner:    {self.score_1} {self.name_1} \nRunner up:     {self.score_2} {self.name_2}'
    
    def __add__(self, other):
        new_tracker = ScoreTracker()

        new_tracker.include(self.score_1, self.name_1)
        new_tracker.include(self.score_2, self.name_2)

        new_tracker.include(other.score_1, other.name_1)
        new_tracker.include(other.score_2, other.name_2)

        return new_tracker


# scores = ScoreTracker()
# scores.include(105, 'Alice')
# print(scores)
# # HIGH SCORES
# # Winner    105 Alice
# # Runner up   0 
# scores.include(98, 'Bob')
# print(scores)
# # HIGH SCORES
# # Winner     98 Bob
# # Runner up 105 Alice
# scores.include(108, 'Alice')
# print(scores)
# # HIGH SCORES
# # Winner    108 Alice
# # Runner up  98 Bob

wednesday_scores = ScoreTracker()
wednesday_scores.include(105, 'Alice')
wednesday_scores.include(98, 'Bob')
print('\nWEDNESDAY', wednesday_scores)

# WEDNESDAY HIGH SCORES
# Winner    105 Alice
# Runner up  98 Bob

thursday_scores = ScoreTracker()
thursday_scores.include(102, 'Charlie')
thursday_scores.include(99, 'Alice')
print('\nTHURSDAY', thursday_scores)

# THURSDAY HIGH SCORES
# Winner    102 Charlie
# Runner up  99 Alice
 
combined_scores = wednesday_scores + thursday_scores
#combined_scores = wednesday_scores.__add__(thursday_scores)

print('\nCOMBINED', combined_scores)

# COMBINED HIGH SCORES
# Winner    105 Alice
# Runner up 102 Charlie
```

## Code 11.3: Unique Score Tracker

```python
class UniqueScoreTracker(ScoreTracker): #nedarver Scoretracker
    def __init__(self):
        super().__init__() #Initierer ingen nye ting
   


    def include(self, score, name):
        if score >= self.score_1:
            if name == self.name_1:
                self.score_1 = score
            else:
                self.score_2 = self.score_1
                self.name_2 = self.name_1

                self.name_1 = name
                self.score_1 = score
                
        elif score >= self.score_2:
            if name != self.name_1:
                self.score_2 = score
                self.name_2 = name
        


scores = UniqueScoreTracker()
scores.include(105, 'Alice')
scores.include(98, 'Bob')
scores.include(108, 'Alice')
print(scores)
# HIGH SCORES
# Winner    108 Alice
# Runner up  98 Bob
scores.include(99, 'Alice')
print(scores)
# HIGH SCORES
# Winner    108 Alice
# Runner up  98 Bob
scores.include(109, 'Bob')
print(scores)
# HIGH SCORES
# Winner    109 Bob
# Runner up 108 Alice   

```
