### Math Functions


```python
import matplotlib.pyplot as plt

# Define n
n = 100

# Create a Numpy array x with n equally spaced points between 0 and 1
x = np.linspace(0, 1, n) 
#print(x)

# Define the function f(x)
y = np.abs(1/2 - x**3) - np.sqrt(np.sin(x) + np.exp(x)) #dette er også et array
#print(y)

# Plot the function
plt.plot(x, y, color='hotpink')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Plot of f(x)')
plt.grid(True)
plt.show()
```
