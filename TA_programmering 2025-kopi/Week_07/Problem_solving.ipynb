### Problem 7.9: DNA Complement

```python
def dna_complement(dna, reverse):
    compliment = ""
    
    for i in dna:
        if i == "A":
            compliment += "T"
        elif i == "T":
            compliment += "A"
        elif i == "C":
            compliment += "G"
        else:
            compliment += "C"
    
    if reverse:
        return compliment[::-1]
    else:
        return compliment

print(dna_complement("AATTCG", True))
```

### Problem 7.10: Generalized Palindrome

```python
def is_palindrome(text):
    new_string = ""

    for i in text.lower():
        if i.isalpha() and i != ' ':
            new_string += i
    
    return new_string == new_string[::-1]

print(is_palindrome("Was it a car or a cat I saw?"))
```

```python
test0 = is_palindrome('racecar')
test1 = is_palindrome('A man, a plan, a canal: Panama!')
test2 = is_palindrome('Was it a car or a cat I saw?')
test3 = is_palindrome('Damejer nis berg greb sin rejemad.')
test4 = is_palindrome('A Santa at NASA.')
test5 = not is_palindrome('Morten')
test6 = not is_palindrome('Not a palindrome')
test7 = is_palindrome('Madam')
test8 = not is_palindrome('Palindromes are cool')
test9 = not is_palindrome('Level and level')
test10 = not is_palindrome('Sequence of letters.')

all_tests_passed = (test0 and test1 and test2 and test3 and test4 and test5 
        and test6 and test7 and test8 and test9 and test10) 
print("All tests passed:", all_tests_passed)
```

### Problem 7.11: Punctuation Check

```python
def punctuation_check(text):
    count = 0
    punc = ",.:?;!"

    for i in range(len(text) - 1):
        if text[i] in punc:
            if text[i + 1] != ' ':
                count += 1

    return count
text = "This is:a text, a text with many,many errors! But,who cares?"
print(punctuation_check(text))
```

```python
print(punctuation_check("This is:a text, a text with many,many errors! But,who cares?"))
print(punctuation_check("Here we have no errors.Or do we?!"))
print(punctuation_check("No,I am sure this is error-free!"))
print(punctuation_check("Whataboutonlyasingleword? "))
print(punctuation_check("       almost   empty    ?     "))
print(punctuation_check("3 times.is.not enough"))
print(punctuation_check("!Whou, this is strange!"))
print(punctuation_check("A,B,C,D,E,F,G,H,I"))
print(punctuation_check("Number nine is a good example, almost no problems there-I think."))
print(punctuation_check("But!The last text is bad.Really,really bad.No space at all.Squeezed!"))
```

### Problem 7.12: Code Shift

```python
def code_shift(code, turn):
    new_string = ""

    for i in code:
        new_digit = int(i) + turn
        if new_digit % 10 >= 0:
            new_digit = new_digit % 10
        
        new_string += str(new_digit)

    return new_string

print(code_shift('2638', 3))
print(code_shift('9165', 1))
```

```python
test0 = code_shift('2638', 3) == '5961'
test1 = code_shift('0120', -3) == '7897'
test2 = code_shift('9165', 1) == '0276'

print(test0, test1, test2)
all_tests_passed = test0 and test1 and test2 
print("All tests passed:", all_tests_passed)
```

### Problem 7.13: Arrival Times 

```python
def arrival_times(scheduled_times, delay):

    new_times = []

    for i in scheduled_times:

        timelist =  i.split(':')

        hour = int(timelist[0])
        minutes =  int(timelist[1])

        total_minutes = minutes + delay
        total_hour = hour + total_minutes // 60

        final_minutes = total_minutes % 60
        final_hour = total_hour % 24

        new_times.append(f'{final_hour:02}:{final_minutes:02}')
    
    return new_times

print(arrival_times(['12:37', '08:10'], 25))
#['13:02', '08:35']
```

```python
returned = arrival_times(['12:37', '08:10'], 25)
expected = ['13:02', '08:35']

if returned != expected:
    print('Test for Arrival Times failed because returned was:')
    print(repr(returned))
    print('instead of:')
    print(repr(expected))
else:
    print('Test for Arrival Times passed')
```

### Problem 7.14: Punctuation Ratio 


```python
def punctuation_ratio(text):
    comma = 0
    no_comma = 0

    text = text.split(' ')

    for i in range(len(text)):
        if text[i] == 'and':
            if text[i - 1][-1] == ',':
                comma += 1
            else:
                no_comma += 1
    return comma/no_comma

text = text = ("Sara and Emma like to travel, bike, and hike, and when they are " + "traveling they always take their bikes, hiking shoes, and sleeping bags. " +
"Last year, Sarah and Emma traveled to Italy, France, and Spain. And that " +
"was fun, and, according to Sara and Emma, very expensive!")

print(punctuation_ratio(text))
```
