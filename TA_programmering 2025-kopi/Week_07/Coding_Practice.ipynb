### Code 7.1: Most Vowels

```python
def vowels(sentence):
    most_vowels = ''
    most_count = 0

    tjek = 'aeouyi'
    for word in sentence.lower().split(' '):
        count = 0
        for letter in word:
            if letter in tjek:
                count += 1
        if count > most_count:
            most_count = count
            most_vowels = word

    return most_vowels


sentence = "Syntax is not sufficient for semantics"

print(vowels(sentence))

```

### Code 7.2: Change Case

```python
def change_case(word):
    words_list = []

    final_words = ""

    for i in range(len(word) - 1):
        if word[i] == "_":
            words_list.append(final_words)
            final_words = ""
            final_words += word[i + 1].upper()
        elif word[i] == word[i].upper():
            words_list.append(final_words)
            final_words = ""
            final_words += "_" + word[i].lower()
        else:
            final_words += word[i]

    return words_list

print(change_case("thisIsIdentifierName"))
print(change_case("student_grade_average"))
```

### Code 7.4: Framed Sign Function

```python
def print_framed_sign(message, border_char):
    samlet = (len(message) + 4) * border_char + "\n| " + message + " |\n" + (len(message) + 4) * border_char
    return samlet

print(print_framed_sign("Harry, did you put your name in the goblet of fire?", "~"))
```

```python
def print_framed_sign(message, border_char, is_important):
    if is_important:
        message = message.upper()
        samlet = (len(message) + 4) * border_char + "\n| " + message + " |\n" + (len(message) + 4) * border_char
        return samlet
    else:
        samlet = (len(message) + 4) * border_char + "\n| " + message + " |\n" + (len(message) + 4) * border_char
        return samlet

print(print_framed_sign("Harry, did you put your name in the goblet of fire?", "~", False))
```

### Code 7.6: Count Vowels

```python
def vowel_count(test):
    vo = "aeiuo"
    counter = 0
    for i in test.lower():
        if i in vo:
            counter += 1
    
    return counter

print(vowel_count("Hello, what's up?"))

print(vowel_count("OMG, I can't believe it!"))

print(vowel_count("This is a day when I'm feeling good"))
```

### Code 7.8: Format Dates

```python
def format_date(date):
    months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'] 

    date = date.split("/")

    new_string = date[0] + "-" + months[int(date[1])] + "-" +  date[2][2:]

    return new_string


new_format = format_date("01/08/2020")
print(new_format)
```
