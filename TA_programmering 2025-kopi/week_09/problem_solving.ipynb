### Problem 9.6 Lix Number

```python
def lix_number(filename):
    with open(filename, "r") as file:
        lines = file.readlines()

    sentence_sum = 0
    total_word = 0
    words = 0
    
    for i in lines:
        sentence_sum += len(i.strip().split())

        for word in i.strip().split():
            total_word += 1
            if len(word.strip(".,!?:;")) > 6:
                words += 1
    
    ave_sentence = sentence_sum / len(lines)
    procentage = (words / total_word) * 100

    if len(lines) == 0:
        return 0
    else:
        return ave_sentence + procentage



print(lix_number('texts/text_lix_a.txt'))
print(lix_number('texts/text_lix_g.txt'))
```

### Problem 9.7: Number Lines (combination af filer er ikke eksamenrelevant)

```python
def number_lines(filename):
    new_lines = ''
    with open(filename, 'r') as file:
        lines = file.readlines()
        new_lines += lines[0] + '\n'
        start_number = 0
        for i in lines[2:]:
            # Check if the line is empty or only contains whitespace
            if i.strip() == '':
                new_lines += '\n'
            else:
                start_number += 1
                # Format line numbers with two spaces, aligning single-digit numbers with a leading space
                new_lines += f'{start_number:2}  {i}'
    
    # Create the new filename by adding '_numbered' before '.txt'
    new_filename = filename.replace('.txt', '_numbered.txt')
    
    # Write the new content to the new file
    with open(new_filename, 'w') as new_file:
        new_file.write(new_lines)

    print(f"The numbered file has been saved as: {new_filename}")

number_lines('number_lines/bro_bro_brille.txt')
```

### Problem 9.8: Nitrate Levels

```python
def nitrate_levels(path):

    with open(path, 'r') as file:
        numbers = file.readlines()
        Very_low = 0
        Low = 0
        Normal = 0
        High = 0
        Very_high = 0
        for i in numbers:
            i = float(i)
            if i <= 4.0:
                Very_low += 1
            elif 4.0 < i <= 9:
                Low += 1
            elif 9 < i < 40.0:
                Normal += 1
            elif 40 <= i < 50.0:
                High += 1
            else:
                Very_high += 1
        
        return Very_low, Low, Normal, High, Very_high

print(nitrate_levels('week_09_files/nitrate_data_A.txt'))
```
