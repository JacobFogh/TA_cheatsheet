### Code 9.1: Surname Percentage

```python
def surname_percentage(path, name):

    with open(path) as file:
        lines = file.readlines()

        summen = 0
        for i in lines: 
            surname, number = i.strip().split(",")
            summen += int(number)
            
        for line in lines:
            surname, number = line.strip().split(",")
            if surname == name:
                return (int(number) * 100)/summen

            



surname_percentage('week_09_files/efternavne.csv', 'Jensen')
```

### Code 9.3: Count a Letter

```python


def count_letter(filename, letter):
    with open(filename, "r") as file:
        content = file.read()

    letter = "a"
    count = 0
    for i in content.lower():
        if i == letter.lower():
            count += 1

    return count

filename = "texts/quick_fox.txt"
print(count_letter(filename, "b"))
```

### Code 9.4: Letter Frequency

```python
def count_letters(filename):

    with open(filename, "r") as file:
        content = file.read()

    new_dict = {}
    for i in content.lower():
        if i not in new_dict:
            new_dict[i] = 1
        else:
            new_dict[i] += 1
    
    return new_dict

filename = "texts/quick_fox.txt"
print(count_letters(filename))
```

### Code 9.5: Combining Files

```python
import os

content_list = []

separator = "\n\n" + "-" * 15 + "\n\n"


folder_path = "number_lines"  

#print(os.listdir(folder_path))

for filename in os.listdir(folder_path): 


    if filename.endswith(".txt"):
        file_path = os.path.join(folder_path, filename) 

    
        with open(file_path, 'r') as file:
            content = file.read()
            content_list.append(content)  
            #print(f"{filename} har længden: {len(content)}") 



combined_content = separator.join(content_list)
print(combined_content)


with open("combined.txt", 'w') as combined_file:
    combined_file.write(combined_content)
```
