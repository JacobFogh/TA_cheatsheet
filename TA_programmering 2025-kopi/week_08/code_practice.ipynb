### Code 8.1: Phonebook Merge

```python
def phonebook_merge(book1, book2):

    for i in book2:
        if i not in book1:
            book1[i] = book2[i]
        else:
            for number in book2[i]:
                if number not in book1[i]:
                    book1[i].append(number)
    
    return book1


phonebook = {'Liv': ['55511112', '18777890'] ,
              'Mads': ['27274445', '48533336'],
              'Steve': ['45455555', '25455525']}

second_phonebook = {'Anna': ['89577772'] ,
                     'Steve': ['25257755', '25455525'],
                     'Mads': ['48533336', '27274445']}

print(phonebook_merge(phonebook, second_phonebook))
```

### Code 8.2: Inventory Management

```python
def restock_inventory(inventory, restock):
    for i in inventory:
        inventory[i] += restock
    
    return inventory
inventory = {
    "apples": 10,
    "bananas": 5,
    "oranges": 8,
    "kiwis": 7}

print(restock_inventory(inventory, 3))
```

### Code 8.3: Weather Station

```python
# 1) Basen
weather_station = {
    "station_name": "CPH Airport Weather Station",
    "is_operational": True,
    "city": "Kastrup",
    "country": "Denmark",
    "construction_year": 2016,
}

# 2) Ekstra data
weather_station["temperature_readings"] = [20.1, 21.3, 19.8, 22.0]
weather_station["wind_speed_readings"] = [5.2, 3.4, 7.1, 15.6]

# 3) Tjek: print alle key–value-par
for key, value in weather_station.items():
    print(f"{key}: {value}")
```

```python
def get_average_readings(weather_station):
    ave_speed = sum(weather_station["wind_speed_readings"]) / len(weather_station["wind_speed_readings"])

    ave_temp = sum(weather_station["temperature_readings"]) / len(weather_station["temperature_readings"])

    return ave_temp, ave_speed


print(get_average_readings(weather_station))
```

```python
import math
def data_statistics(values, name):

    mean = (sum(values) / len(values))
    summen = 0
    for i in values:
        summen += (i - mean)**2
    
    std = math.sqrt(summen / len(values))

    std2 = math.sqrt(sum([(x - mean)**2 for x in values]) / len(values))

    print(std, std2)

    return  {"experiment_name": name, 
            "experiment_values": values, 
            "mean_val": mean,
            "standard_deviation": std, 
            "max_val": max(values), 
            "min_val": min(values)}

experiment_name = "standard_values"
experiment_values = [1.30, -0.01,  1.35,  0.31,  0.35, -1.61,  1.39,  1.19, -1.07,  0.34]

print(data_statistics(experiment_values, experiment_name))
    
```
