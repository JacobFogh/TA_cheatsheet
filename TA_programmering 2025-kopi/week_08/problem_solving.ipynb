### Problem 8.5: Energy Expenditure

```python
def calculate_bee_tee(person_data, name):
    
    person_index = person_data["names"].index(name)
   
    if person_data["genders"][person_index] == "Male":
        basal = 66 + (13.7 * person_data["weights"][person_index]) + (5 * person_data["heights"][person_index]) - (6.8 * person_data["ages"][person_index])
    
    else:
        basal = 655 + (9.6 * person_data["weights"][person_index]) + (1.8 * person_data["heights"][person_index]) - (4.7 * person_data["ages"][person_index])

    if person_data["activity_levels"][person_index] == "moderate":
        AF = 1.5
    elif person_data["activity_levels"][person_index] == "strenuous":
        AF = 2.0
    else: 
        AF = 1.3
    
    
    return (basal, (basal * AF)) 


person_data = {
    "names": ["Jakob", "Josefine", "Morten"],
    "genders": ["Male", "Female", "Male"], 
    "ages": [25, 30, 37], 
    "weights": [70, 60, 86], 
    "heights": [175, 165, 189],
    "activity_levels": ["moderate", "strenuous", "sedentary"]
}

print(calculate_bee_tee(person_data, 'Jakob'))

print(calculate_bee_tee(person_data, 'Josefine'))

print(calculate_bee_tee(person_data, 'Morten'))
```

### Problem 8.6: Player Scores

```python
def get_highscore(player_stats):
    highest_score = 0
    highest_score_name = ""

    for player in player_stats:
        score = 0

        score += player["goals"] * 20
        score += player["passes"] * 2
        score += (player["goal_attempts"] - player["goals"]) * 5
        score += (player["pass_attempts"] - player["passes"]) * 1

        if score > highest_score:
            highest_score = score
            highest_score_name = player["name"]

    return highest_score_name, highest_score


player_stats = [
    {"name": "Bob", "goals": 2, "passes": 15, "goal_attempts": 3, "pass_attempts": 17},
    {"name": "Alex", "goals": 1, "passes": 17, "goal_attempts": 2, "pass_attempts": 20},
    {"name": "Molly", "goals": 1, "passes": 28, "goal_attempts": 2, "pass_attempts": 31},
    {"name": "Sophia", "goals": 2, "passes": 21, "goal_attempts": 3, "pass_attempts": 25}
]

print(get_highscore(player_stats))
```

### Problem 8.7: Social Network

```python
def popularity(social_network, person_name):
    tjek = social_network[person_name]

    mutual = 0
    outgoing = 0
    ingoing = 0

    for names in social_network:
        if (names in tjek) and (person_name in social_network[names]):
            mutual += 1
        elif (names in tjek) and (person_name not in social_network[names]):
            outgoing += 1
        elif (names not in tjek) and (person_name in social_network[names]):
            ingoing += 1
        

    return mutual, outgoing, ingoing

social_network = {
    "Alice": ["Josefine", "Charlie", "Jakob"],
    "Morten": ["Charlie", "David", "Josefine"],
    "Josefine": ["Alice", "David", "Jakob"],
    "Charlie": ["Alice", "Jakob", "David"],
    "David": ["Josefine"],
    "Jakob": ["Charlie", "Alice", "Josefine", "David"]
}

print(popularity(social_network, "Charlie"))
```

### Problem 8.8: Booklet Layout

```python
def booklet_layout(content_pages):
    blank = 0

    while content_pages % 4 != 0:
        blank += 1
        content_pages += 1
    
    return content_pages, blank

print(booklet_layout(17))
```

```python
returned = booklet_layout(17)
expected = (20, 3)

if returned != expected:
    print('Test for Booklet Layout failed because returned was:')
    print(repr(returned))
    print('instead of:')
    print(repr(expected))
else:
    print('Test for Booklet Layout passed')
```

### Problem 8.9: Name Frequency

```python
def name_frequency(names):
    new_dict = {}

    for i in names:
        first_name = i.split()[0]
        if first_name not in new_dict:

            new_dict[first_name] = 1
        else:
            new_dict[first_name] += 1
    
    return new_dict

names = ['Liv Ea Jensen', 
         'Mads Oliver', 
         'Steve Madsen',
         'Anna Simon', 
         'Simon Gade', 
         'Mads Kai Jensen']

print(name_frequency(names))
```
