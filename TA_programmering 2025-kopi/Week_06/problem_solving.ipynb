## Problem 6.11: Unique Values

```python
def unique_values(RNA):
    unique = []

    for i in RNA:
        if i not in unique:
            unique.append(i)
    return unique

RNA = ['A', 'G', 'G', 'C', 'U', 'U', 'A', 'A', 'U', 'C', 'C', 'G']
print(unique_values(RNA))

print(unique_values(['Fr', 'Fr', 'Sa', 'Su', 'Sa', 'Sa', 'Su', 'Su', 'Mo', 'Fr', 'Sa', 'Fr']))

print(unique_values([1, 2, 20, 6, 210, 20, 2]))
```

## Problem 6.12: Best Buy

```python
def best_buy(prices, budget):
    count = 0

    for i in prices:
        if budget >= i:
            count += 1
            budget -= i
        else:
            break
    
    return count

print(best_buy([5, 4, 6, 2, 9, 1, 1, 4], 16))
```

## Problem 6.13: Strategic Buy


```python
def strategic_buy(prices, budget):

    prices = sorted(prices)
    count = 0

    for i in prices:
        if budget >= i:
            count += 1
            budget -= i
    
    return count
```

```python
expected = 3
prices = [1, 12, 5, 14, 1000]
output = strategic_buy(prices, 30)
if output != expected:
    print("FAILED the following test:")
print("strategic_buy(", prices, ", 30) returned", output, "should be", expected)

expected = 0
prices = [11, 2 ,5, 7, 3]
output = strategic_buy(prices, 1)
if output != expected:
    print("FAILED the following test:")
print("strategic_buy(", prices, ", 1) returned", output, "should be", expected)

expected = 9
prices = [4, 10, 7, 9, 6, 15, 13, 12, 1, 5, 14, 11, 3, 8, 2]
output = strategic_buy(prices, 45)
if output != expected:
    print("FAILED the following test:")
print("strategic_buy(", prices, ", 45) returned", output, "should be", expected)

expected = 0
prices = [] # nothings for sale
output = strategic_buy(prices, 100)
if output != expected:
    print("FAILED the following test:")
print("strategic_buy(", prices, ", 100) returned", output, "should be", expected)
```

## Problem 6.14: Duplicate Elements

```python
def duplicate_elements(elements, dup):
    dupli = []

    for i in elements:
        for n in range(dup):

            dupli.append(i)
    
    return dupli

print(duplicate_elements([4, "B", 3.9], 3))
```

## Problem 6.15: Water Height

```python
def water_height(h0, rain_per_day):
    

    for i in rain_per_day:
        today = h0 + i - 2
        h0 = today
    if today < 0:
        today = 0
        
    return today
    



print(water_height(0, [1]))

print(water_height(0, [4, 2, 3]))

print(water_height(5, [1, 0, 2, 1]))
```

## Problem 6.16: Special Occurrence

```python
def special_occurrence(my_list):
    for i in range(len(my_list) - 1):
        if my_list[i] == 5:
            if (my_list[i + 1] == 7 and my_list[i + 2]) != 7:
                return i
            elif (my_list[i + 1] != 7 and my_list[i + 2]) == 7:
                return i



print(special_occurrence([2, 8, 11, 3, 12, 5, 7, 7, 11, 3, 12, 5, 2, 7, 5, 7, 2, 6]))
```

```python
returned = special_occurrence([2, 8, 11, 3, 12, 5, 7, 7, 11, 3, 12, 5, 2, 7, 5, 7, 2, 6])
expected = 11

if returned != expected:
    print('Test for Special Occurrence failed because returned was:')
    print(repr(returned))
    print('instead of:')
    print(repr(expected))
else:
    print('Test for Special Occurrence passed')
```

## Problem 6.17: First Double Peak

```python
def first_double_peak(sequence):

    for i in range(1, len(sequence)):
        if sequence[i - 2] < sequence[i] and sequence[i - 1] < sequence[i]:
            if sequence[i + 2] < sequence[i] and sequence[i + 1] < sequence[i]:
                return i
    
    return -1


print(first_double_peak([1.2, 2.4, 3.1, 2.9, 3.6, 2.3, 1.9, 2.4]))
```
