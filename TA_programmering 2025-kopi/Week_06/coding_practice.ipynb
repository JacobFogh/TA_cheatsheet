## Code 6.2 State Durations

```python
def state_function(states, stamps):

    state_0 = 0
    state_1 = 0
    state_2 = 0

    for i in range(len(states) - 1):
        if states[i] == 0:
            state_0 += abs(stamps[i] - stamps[i + 1])
            print(state_0)
        elif states[i] == 1:
            state_1 += abs(stamps[i] - stamps[i + 1])
            print(state_1)
        else:
            state_2 += abs(stamps[i] - stamps[i + 1])
            print(state_2)
    
    return [state_0, state_1, state_2]


states = [2,1,2,0]
stamps = [0, 200, 500, 600, 800]
print(state_function(states, stamps))
```

### Code 6.4: Powers of 2

```python
def power_of_two(n):
    numbers = []

    for i in range(1, n+1):
        numbers.append(2**i)

    return numbers



print(power_of_two(4))

```

>En smart måde og gøre det samme på, er gjort under:

```python
def power_of_two(n):
    return [2**i for i in range(1, n+1)]  #Dette er en lidt mere advanceret men kort måde og gøre det på
```

## Code 6.5: Average of a List

```python
def average(numbers):
    gennemsnit = sum(numbers)/len(numbers)
    return gennemsnit

print(average([2, 4, 8, 16]))
```

## Code 6.6: Slicing the Months

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

spring = months[2:5]
print(spring)

summer = months[6:9]
print(summer)

quarter_starts = months[2::3]
print(quarter_starts)

reversed_months = months[::-1]
print(reversed_months)

winter = months[-1:] + months[:2]
print(winter)
```

## Code 6.7: Long Months

```python
def long_months(months, days, d):
    long = []

    for i in range(len(months) - 1):
        if days[i] > d:
            long.append(months[i])
    
    return long

    #return [months[i] for i in range(len(months) - 1) if days[i] > d] 

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
days = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
d = 30

print(long_months(months, days,d))
```

## Code 6.9: Top k Elements

```python
def top_k(integers, k):
    sorted_list = sorted(integers)
    return sorted_list[len(integers) - k:]

print(top_k([0, -5, 2], 1))
print(top_k([6, 6, 2, 25, 1, 999, -5], 3))
```

## Code 6.10: Is Available?

```python
def which_available(menu, desired_dishes):
    available = []

    for i in desired_dishes:
        if i in menu:
            available.append(True)
        else:
            available.append(False)
    return available


menu = ["pasta", "pizza", "salad", "soup", "steak", "chicken", "sandwich"]
dishes = ["salad", "beer", "pasta", "hotdog", "dog"]
print(which_available(menu, dishes))
#[True, False, True, False, False]
print(which_available([], ["salad"]))
#[False]
```
